## 2. Mechanism and math

A rigid transform combines rotation and translation:
\[
p_A = R_{AB}p_B + t_{AB}
\]
In homogeneous form:
\[
[p_A; 1] = T_{AB} [p_B; 1]   (homogeneous coordinates)
\]
Composition chains frames: world ← base ← camera ← point.

## 3. From-scratch lab

Use the repo transform utilities to move a point between frames.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import numpy as np
from robotics.transforms import Rz, make_transform, transform_points, invert_transform
T_world_from_robot = make_transform(Rz(np.pi/2), [1, 0, 0])
point_robot = np.array([[1, 0, 0]])
point_world = transform_points(T_world_from_robot, point_robot)
print('point in robot frame:', point_robot)
print('point in world frame:', np.round(point_world, 3))
print('back to robot:', np.round(transform_points(invert_transform(T_world_from_robot), point_world), 3))
```
