## 2. Mechanism and math

A simple VLA policy has the shape:
\[
a_t = \pi_\theta(o^{image}_t, instruction, state_t)
\]
Training often begins with behavioral cloning:
\[
\min_\theta \|\pi_\theta(o, text)-a^{expert}\|^2
\]
or cross-entropy for discrete actions.
