## 2. Mechanism and math

A* searches for a low-cost path using:
\[
f(n)=g(n)+h(n)
\]
where `g` is cost so far and `h` is a heuristic estimate to the goal. Good heuristics speed search without changing the answer if admissible.

## 3. From-scratch lab

Plan around obstacles with A*.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import numpy as np
from robotics.planning import astar
grid = np.zeros((6,6), dtype=int)
grid[2,1:5] = 1
start, goal = (0,0), (5,5)
path = astar(grid, start, goal)
print('path:', path)
for r in range(grid.shape[0]):
    print(''.join('S' if (r,c)==start else 'G' if (r,c)==goal else '*' if (r,c) in path else '#' if grid[r,c] else '.' for c in range(grid.shape[1])))
```
