## 2. Mechanism and math

Behavioral cloning is supervised learning:
\[
\min_\theta \sum_i \ell(\pi_\theta(o_i), a_i^{expert})
\]
Its main failure is covariate shift: small mistakes lead to states not seen in demonstrations.

## 3. From-scratch lab

Create a toy expert and clone its action rule.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
# Expert rule: if position is left of target, move right; if right, move left.
data = []
for x in [-2,-1,0,1,2]:
    action = 1 if x < 0 else -1 if x > 0 else 0
    data.append((x, action))
print('demonstrations:', data)
print('BC means fit a model f(x)->action on these examples.')
```
