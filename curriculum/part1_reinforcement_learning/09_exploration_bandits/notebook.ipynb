## 2. Mechanism and math

Regret measures reward lost compared with always choosing the best arm:
\[
\text{regret}_t = \mu^* - \mu_{a_t}
\]
Good exploration minimizes cumulative regret while still discovering which action is best.

## 3. From-scratch lab

Compare epsilon-greedy and UCB on a tiny Bernoulli bandit.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
from rl.agents.bandits import BernoulliBandit, EpsilonGreedy, UCB1, run_bandit
bandit = BernoulliBandit([0.2, 0.5, 0.8], seed=0)
for agent in [EpsilonGreedy(3, eps=0.1), UCB1(3)]:
    rewards, regret = run_bandit(bandit, agent, steps=1000)
    print(type(agent).__name__, 'avg reward', round(sum(rewards)/len(rewards),3), 'regret', round(regret[-1],1))
```
