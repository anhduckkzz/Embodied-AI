## 2. Mechanism and math

Bellman thinking says: the value of now equals immediate reward plus discounted value of next state:
\[
V(s)=\mathbb{E}[r+\gamma V(s')]
\]
This recursive idea powers dynamic programming, Q-learning, DQN, actor-critic, and planning.

## 3. From-scratch lab

Compute discounted return by hand. This is what every value target approximates.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
rewards = [1, 1, 1, 10]
gamma = 0.9
G = 0
for r in reversed(rewards):
    G = r + gamma * G
    print('partial return:', round(G, 3))
print('G_0 =', round(G, 3))
```
