## 2. Mechanism and math

A constrained RL objective separates reward from cost:
\[
\max_\pi \mathbb{E}[\sum r_t] \quad \text{s.t.}\quad \mathbb{E}[\sum c_t] \le d
\]
Sim-to-real adds robustness: train over varied simulated worlds so the policy survives reality.

## 3. From-scratch lab

Domain randomization: sample many worlds instead of one perfect simulator.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import random
for episode in range(5):
    mass = random.uniform(0.8, 1.2)
    friction = random.uniform(0.5, 1.5)
    sensor_noise = random.uniform(0.0, 0.05)
    print(f'episode {episode}: mass={mass:.2f}, friction={friction:.2f}, noise={sensor_noise:.3f}')
```
