## 2. Mechanism and math

A hierarchical policy can be written as high-level option selection plus low-level control:
\[
z_t \sim \pi_{high}(z|s), \quad a_t \sim \pi_{low}(a|s,z)
\]
This reduces long-horizon complexity.

## 3. From-scratch lab

A tiny hierarchy: high-level chooses a target; low-level moves toward it.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
pos = 0
targets = {'go_left': -3, 'go_right': 3}
skill = 'go_right'
for t in range(10):
    target = targets[skill]
    action = 1 if target > pos else -1 if target < pos else 0
    pos += action
    print('t', t, 'skill', skill, 'action', action, 'pos', pos)
    if pos == target: break
```
