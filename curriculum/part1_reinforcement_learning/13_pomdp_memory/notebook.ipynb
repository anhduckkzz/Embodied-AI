## 2. Mechanism and math

In a POMDP, the policy sees observation \(o_t\), not full state \(s_t\). Memory summarizes history:
\[
h_t=f(h_{t-1}, o_t, a_{t-1})
\]
Filters and recurrent networks are two ways to build memory.

## 3. From-scratch lab

A simple hidden-state example: two observations are needed to infer velocity.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
positions = [0, 1, 2, 3]
last = None
for obs in positions:
    if last is None:
        print('obs', obs, 'velocity unknown')
    else:
        print('obs', obs, 'estimated velocity', obs-last)
    last = obs
```
