## 2. Mechanism and math

Value iteration applies the Bellman optimality backup:
\[
V_{k+1}(s)=\max_a \sum_{s'} P(s'|s,a)[R(s,a,s')+\gamma V_k(s')]
\]
If the model is correct and the state/action spaces are small, this converges to the optimal value.

## 3. From-scratch lab

Use the repo implementation on a tiny chain MDP.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import numpy as np
from rl.agents.dynamic_programming import value_iteration
P = np.zeros((3, 2, 3)); R = np.zeros((3, 2, 3))
# action 0 stays, action 1 moves right. Reaching state 2 gives reward 1.
for s in range(3):
    P[s,0,s] = 1
    ns = min(s+1, 2)
    P[s,1,ns] = 1
    if ns == 2: R[s,1,ns] = 1
V, pi = value_iteration(P, R, gamma=0.9)
print('V:', np.round(V, 3))
print('policy:', pi, '(1 means move right)')
```
