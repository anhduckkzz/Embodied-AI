## 2. Mechanism and math

Dyna-Q mixes real and imagined updates:
\[
Q(s,a) \leftarrow Q(s,a)+\alpha[r+\gamma\max_{a'}Q(s',a')-Q(s,a)]
\]
The difference is that many updates come from a learned model instead of the real environment.

## 3. From-scratch lab

Train Dyna-Q on a simple chain where planning accelerates learning.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
from rl.agents.dyna import DynaQ
agent = DynaQ(n_states=5, n_actions=2, planning_steps=10, seed=0)
for episode in range(50):
    s = 0
    for t in range(20):
        a = agent.act(s)
        ns = min(s+1, 4) if a == 1 else s
        r = 1.0 if ns == 4 else 0.0
        done = ns == 4
        agent.step(s, a, r, ns, done)
        s = ns
        if done: break
print('Q at start:', agent.Q[0], 'best action:', agent.Q[0].argmax())
```
