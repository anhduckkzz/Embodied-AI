## 2. Mechanism and math

Pinhole projection maps a 3D point to pixels:
\[
u=f_x X/Z+c_x, \quad v=f_y Y/Z+c_y
\]
Depth `Z` is the missing dimension in a single RGB image.

## 3. From-scratch lab

Project 3D points into a camera image.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
points = [(0,0,2), (1,0,2), (0,1,4)]
fx, fy, cx, cy = 500, 500, 320, 240
for X,Y,Z in points:
    u = fx*X/Z + cx
    v = fy*Y/Z + cy
    print((X,Y,Z), '-> pixel', (round(u,1), round(v,1)))
```
