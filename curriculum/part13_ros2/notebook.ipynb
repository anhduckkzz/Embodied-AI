## 2. Mechanism and math

ROS is message passing. A node publishes typed messages on topics; other nodes subscribe. Transforms are tracked by tf2; robot structure is described by URDF.

## 3. From-scratch lab

Simulate the idea of publish/subscribe without installing ROS.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
class Topic:
    def __init__(self): self.subscribers = []
    def subscribe(self, fn): self.subscribers.append(fn)
    def publish(self, msg):
        for fn in self.subscribers: fn(msg)

def listener(msg): print('listener received:', msg)
chatter = Topic()
chatter.subscribe(listener)
chatter.publish({'data': 'hello robot'})
```
