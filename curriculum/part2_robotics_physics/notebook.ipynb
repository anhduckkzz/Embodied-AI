## 2. Mechanism and math

Newton/Euler intuition: force changes linear motion and torque changes angular motion.
\[
F=ma, \quad \tau=I\alpha
\]
A simulator numerically integrates these equations over small time steps. Controllers and RL policies output commands, but physics decides the next state.

## 3. From-scratch lab

Simulate a 1D mass pushed by force. Notice that action changes acceleration first, then velocity, then position.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
dt = 0.02
mass = 2.0
x, v = 0.0, 0.0
for step in range(10):
    force = 1.0
    a = force / mass
    v += a * dt
    x += v * dt
    print(f'step={step} a={a:.2f} v={v:.3f} x={x:.3f}')
```
