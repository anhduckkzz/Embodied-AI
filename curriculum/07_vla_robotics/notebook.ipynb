## Run the tiny VLA: one policy, two instructions
The same observation demands different actions depending on the language instruction — and a single conditioned policy learns both via behavioral cloning.

```python
import sys; sys.path.insert(0, 'curriculum/07_vla_robotics')
import vla_minidemo as demo
import torch; torch.manual_seed(0)

policy = demo.ConditionedPolicy()
demo.train_bc(policy, epochs=60)
acc = demo.evaluate(policy)
print(f'\nInstruction-following success: {acc:.1%}')
demo.demonstrate_conditioning(policy)
```

### What to change to make it a *real* VLA
- Replace the `(x,y)` observation with **camera images** + a vision encoder (ViT/SigLIP).
- Replace the instruction `Embedding` with a **text encoder** (a real tokenizer/LLM).
- Replace the grid with a **robot env** and continuous actions (Module 05).
- **Scale the demonstration data**, then **RL-fine-tune** with PPO (Module 04).

That path — BC at scale + RL refinement — is how RT-2, OpenVLA, Octo and π0 are built.
