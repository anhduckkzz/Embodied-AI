## 2. Mechanism and math

A sensor measurement is usually modeled as:
\[
z = h(x) + \epsilon
\]
where `x` is true state, `h` is the measurement function, and noise `epsilon` makes the reading imperfect.

## 3. From-scratch lab

Simulate range measurements from a 2D LiDAR-like sensor.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import numpy as np
from robotics.sensors import simulate_lidar_2d, ranges_to_points
obstacles = np.array([[2.0, 0.0], [1.0, 1.0], [-1.0, 0.5]])
angles, ranges = simulate_lidar_2d(obstacles, n_beams=9, max_range=3.0)
points = ranges_to_points(angles, ranges)
print('angles:', np.round(angles,2))
print('ranges:', np.round(ranges,2))
print('points:\n', np.round(points,2))
```
