## 2. Mechanism and math

The repeated pattern is tokenization + embedding + sequence model:
\[
tokens \rightarrow embeddings \rightarrow transformer \rightarrow task head
\]
For VLA, the task head predicts actions instead of only text labels.

## 3. From-scratch lab

Inspect how a sequence of tokens becomes tensors.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
text = 'pick red block'
vocab = {ch:i for i,ch in enumerate(sorted(set(text)))}
tokens = [vocab[ch] for ch in text]
print('vocab:', vocab)
print('tokens:', tokens)
print('A transformer would map each token id to an embedding vector, then mix context with attention.')
```
