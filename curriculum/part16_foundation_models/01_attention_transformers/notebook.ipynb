## 2. Mechanism and math

Scaled dot-product attention:
\[
Attention(Q,K,V)=softmax(QK^T/\sqrt{d})V
\]
Queries ask, keys match, values provide information. The softmax weights are a learned routing table for information.

## 3. From-scratch lab

Run the repo attention implementation and inspect shapes.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import torch
from dl.attention import scaled_dot_product_attention, causal_mask
B,H,T,D = 1,2,4,8
q = torch.randn(B,H,T,D); k = torch.randn(B,H,T,D); v = torch.randn(B,H,T,D)
out, weights = scaled_dot_product_attention(q,k,v,mask=causal_mask(T))
print('out shape:', out.shape)
print('attention weights for head 0:\n', weights[0,0].round(decimals=2))
```
