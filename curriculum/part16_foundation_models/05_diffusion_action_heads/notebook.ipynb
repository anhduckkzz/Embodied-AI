## 2. Mechanism and math

Forward diffusion adds noise:
\[
x_t=\sqrt{\alpha_t}x_0+\sqrt{1-\alpha_t}\epsilon
\]
The model learns to predict/remove noise conditioned on observations.

## 3. From-scratch lab

Add different noise levels to a 1D action trajectory.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import math, random
trajectory = [math.sin(t/5) for t in range(20)]
for noise_scale in [0.1, 0.5, 1.0]:
    noisy = [x + random.gauss(0, noise_scale) for x in trajectory]
    print('noise', noise_scale, 'first five:', [round(v,2) for v in noisy[:5]])
```
