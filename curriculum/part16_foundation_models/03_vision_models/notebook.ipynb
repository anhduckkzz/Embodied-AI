## 2. Mechanism and math

ViT patching: an image is divided into non-overlapping patches. Each patch is projected to an embedding, then positional information is added:
\[
image \rightarrow patches \rightarrow embeddings + positions \rightarrow transformer
\]

## 3. From-scratch lab

Create patch embeddings with the repo ViT component.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import torch
from dl.vision import PatchEmbedding
patch = PatchEmbedding(image_size=32, patch_size=8, in_channels=3, dim=64)
imgs = torch.randn(2,3,32,32)
tokens = patch(imgs)
print('patch tokens:', tokens.shape)  # batch, number_of_patches, dim
```
