## 2. Mechanism and math

LoRA writes a weight update as low rank:
\[
W' = W + BA
\]
where `A` and `B` are small trainable matrices. The big base weight `W` stays frozen.

## 3. From-scratch lab

Count parameters for full fine-tuning vs LoRA.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
in_dim, out_dim, rank = 1024, 1024, 8
full = in_dim * out_dim
lora = rank * in_dim + out_dim * rank
print('full trainable params:', full)
print('LoRA trainable params:', lora)
print('reduction:', round(full / lora, 1), 'x fewer')
```
