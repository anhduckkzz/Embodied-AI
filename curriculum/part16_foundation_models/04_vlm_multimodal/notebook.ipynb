## 2. Mechanism and math

CLIP-style contrastive learning brings matching image/text pairs close:
\[
L = CE(image\rightarrow text)+CE(text\rightarrow image)
\]
Cross-attention instead lets one stream directly read from another stream.

## 3. From-scratch lab

Compute CLIP-style loss on fake paired features.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
import torch
from dl.multimodal import DualEncoder, clip_contrastive_loss
model = DualEncoder(image_feat_dim=8, text_feat_dim=6, embed_dim=4)
image_feats = torch.randn(5,8)
text_feats = torch.randn(5,6)
img_emb, txt_emb = model(image_feats, text_feats)
loss = clip_contrastive_loss(img_emb, txt_emb)
print('contrastive loss:', float(loss))
```
