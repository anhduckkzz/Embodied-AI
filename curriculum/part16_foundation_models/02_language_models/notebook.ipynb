## 2. Mechanism and math

Autoregressive LM objective:
\[
\max \sum_t \log p(x_t|x_{<t})
\]
The model learns to predict the next token from previous tokens.

## 3. From-scratch lab

Use the repo character tokenizer to see text become token IDs.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
from dl.tokenizer import CharTokenizer
text = 'robot go left'
tok = CharTokenizer(text)
ids = tok.encode(text)
print('ids:', ids)
print('decoded:', tok.decode(ids))
print('vocab size:', tok.vocab_size)
```
