## 2. Mechanism and math

Autoregressive LM objective:
\[
\max \sum_t \log p(x_t|x_{<t})
\]
The model learns to predict the next token from previous tokens.

## 3. From-scratch lab

Use the repo character tokenizer to see text become token IDs.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
from dl.tokenizer import CharTokenizer
text = 'robot go left'
tok = CharTokenizer(text)
ids = tok.encode(text)
print('ids:', ids)
print('decoded:', tok.decode(ids))
print('vocab size:', tok.vocab_size)
```

## 4. Framework/practice view

Frameworks: Hugging Face `AutoTokenizer` and `AutoModelForCausalLM` load pretrained tokenizers/models. PEFT/LoRA adapts them cheaply.

The goal is not to replace understanding with APIs. The goal is to recognize the same mechanism when a library hides the details.

```python
try:
    from transformers import AutoTokenizer
    tokenizer = AutoTokenizer.from_pretrained('gpt2')
    print(tokenizer('robot go left')['input_ids'][:10])
except Exception as e:
    print('Install transformers / allow model download for this cell:', e)
```
