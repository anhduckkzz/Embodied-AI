## 2. Mechanism and math

A simulation loop repeats:
\[
\text{observe state} \rightarrow \text{choose action} \rightarrow \text{integrate physics} \rightarrow \text{return next state/reward}
\]
The time step `dt` controls numerical accuracy and speed.

## 3. From-scratch lab

Build the smallest simulator: a point mass with a bounded acceleration command.

Read every line. The code avoids clever abstractions so you can see the mechanism.

```python
dt = 0.05
x, v = 0.0, 0.0
for step in range(20):
    target = 1.0
    action = max(-1, min(1, 4*(target - x) - 1*v))  # simple controller
    v += action * dt
    x += v * dt
    print(f'{step:02d} action={action:+.2f} x={x:.3f} v={v:.3f}')
```

## 3.1 Code reading guide

3. **Update rule:** which line is the core mechanism from the math section?
